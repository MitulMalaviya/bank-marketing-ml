### The Sample Data — laptop.csv

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
import warnings
import matplotlib.pyplot as plt
warnings.filterwarnings('ignore')

df_laptop = pd.read_csv('laptop.csv')
df_laptop
```

### Fit the Model Using scikit-learn

```python
# Feature and Target
X = df_laptop[['number']]
y = df_laptop['Price']
```

```python
# Create model
model = LinearRegression()
```

```python
# Train model
model.fit(X, y)
```

```python
# Print parameters
print("Slope (m):", model.coef_[0])
print("Intercept (c):", model.intercept_)
```

```python
# Predict Price for Laptop Number 8
prediction = model.predict([[8]])
print("Predicted Price:", prediction[0])
```

```python
plt.scatter(X,y)
plt.xlabel("Model Number")
plt.ylabel("price")
plt.title("macbook dataset")
plt.show()
```

```python
plt.scatter(X,y,color='blue')

plt.plot(X,model.predict(X),color='red')

plt.xlabel("Model Number")

plt.ylabel("Price")

plt.title("Linear Regression")

plt.show()
```

### Implement Gradient Descent by Hand

```python
import numpy as np

X = df_laptop['number'].values.astype(float)
Y = df_laptop['Price'].values.astype(float)

n = len(X)

# Initial values
m = 0
c = 0

# Learning Rate
lr = 0.001

# Number of Epochs
epochs = 20000

loss_history = []

for i in range(epochs):

    # Prediction
    Y_pred = m * X + c

    # Error
    error = Y_pred - Y

    # Loss (MSE)
    loss = (1/n) * np.sum(error ** 2)
    loss_history.append(loss)

    # Gradients
    dm = (2/n) * np.sum(error * X)
    dc = (2/n) * np.sum(error)

    # Update
    m = m - lr * dm
    c = c - lr * dc

    if i % 4000 == 0:
        print(f"Epoch {i}: Loss={loss:.2f}, m={m:.2f}, c={c:.2f}")

print("\nFinal Slope (m):", m)
print("Final Intercept (c):", c)

# Prediction
pred = m * 8 + c
print("Predicted Price for Laptop Number 8:", pred)
```

### Full Gradient Descent Code (iPhone Example)

```python
print("Sklearn Slope :", model.coef_[0])
print("Gradient Descent Slope :", m)

print("Sklearn Intercept :", model.intercept_)
print("Gradient Descent Intercept :", c)
```

### Charts

```python
import matplotlib.pyplot as plt

plt.plot(loss_history)
plt.title("Loss vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.grid(True)
plt.show()
```
